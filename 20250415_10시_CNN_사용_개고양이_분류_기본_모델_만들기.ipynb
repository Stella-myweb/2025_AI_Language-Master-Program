```python
# * CNN 기반 개/고양이 분류 기본 모델 만들기

import os
import random
import numpy as np
from PIL import Image

# 데이터 경로 설정
base_dir = '/content/drive/MyDrive/Colab Notebooks/dogs-vs-cats/train/train/train'
dog_dir = os.path.join(base_dir, 'dog')
cat_dir = os.path.join(base_dir, 'cat')

# 하이퍼파라미터 설정
IMG_SIZE = (150, 150)  # 이미지 크기
SAMPLE_SIZE = 1250  # 각 클래스의 샘플 수 (10%)

# 데이터셋 준비 함수
def load_images_from_folder(folder, label, sample_size):
   images = []
   filenames = os.listdir(folder)
   random.shuffle(filenames)  # 파일 이름을 무작위로 섞음
   for filename in filenames[:sample_size]:  # 10% 샘플 추출
       img_path = os.path.join(folder, filename)
       img = Image.open(img_path).resize(IMG_SIZE)
       img_array = np.array(img) / 255.0  # 이미지 정규화
       images.append((img_array, label))
   return images

# 개와 고양이 데이터 불러오기
dog_images = load_images_from_folder(dog_dir, 1, SAMPLE_SIZE)
cat_images = load_images_from_folder(cat_dir, 0, SAMPLE_SIZE)

# 데이터 합치기 및 섞기
data = dog_images + cat_images
random.shuffle(data)

# 이미지와 레이블 분리
X, y = zip(*data)
X = np.array(X)
y = np.array(y)
```
